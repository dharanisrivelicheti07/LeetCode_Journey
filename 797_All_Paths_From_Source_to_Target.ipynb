### 797. All Paths From Source to Target
###### 

### Description
###### 

### Java Solution
```java
class Solution {
    public List<List<Integer>> allPathsSourceTarget(int[][] graph) {
        List<List<Integer>> result = new ArrayList<>();
        List<Integer> path = new ArrayList<>();
        path.add(0);
        dfs(0,graph,path,result);
        return result; 
    }

    private void dfs(int node,int[][] graph,List<Integer> path,List<List<Integer>> result){
        if(node == graph.length-1){
            result.add(new ArrayList<>(path));
            return;
        }
            for(int neighbor : graph[node]){
                path.add(neighbor);
                dfs(neighbor,graph,path,result);
                path.remove(path.size()-1);
            }
    }
}
```

### Cpp Solution
```cpp
class Solution {
public:
    vector<vector<int>> result;

    void dfs(int node, vector<vector<int>>& graph, vector<int>& path) {

        
        if (node == graph.size() - 1) {
            result.push_back(path);
            return;
        }

        // Visit all neighbors
        for (int neighbor : graph[node]) {
            path.push_back(neighbor);      // Choose
            dfs(neighbor, graph, path);    // Explore
            path.pop_back();               // Backtrack
        }
    }

    vector<vector<int>> allPathsSourceTarget(vector<vector<int>>& graph) {

        vector<int> path;
        path.push_back(0);

        dfs(0, graph, path);

        return result;
    }
};
```

### Python solution

````python
class Solution:
    def allPathsSourceTarget(self, graph: List[List[int]]) -> List[List[int]]:

        result = []
        path = [0]

        def dfs(node):

            
            if node == len(graph) - 1:
                result.append(path[:])      # Store a copy
                return

           
            for neighbor in graph[node]:
                path.append(neighbor)       # Choose
                dfs(neighbor)               # Explore
                path.pop()                  # Backtrack

        dfs(0)

        return result
````
