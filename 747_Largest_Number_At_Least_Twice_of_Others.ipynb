### 747. Largest Number At Least Twice of Others
#### 

### Procedure
#### 

### Java Solution
```java
class Solution {
    public int dominantIndex(int[] nums) {
        int max = -1;
        int secondmax = -1;
        int maxIndex = -1;
        for(int i=0;i<nums.length;i++){
            if(nums[i]>max){
                secondmax = max;
                max = nums[i];
                maxIndex = i;
            }
            else if(nums[i] > secondmax){
                secondmax = nums[i];
            }
        }
        if(max >= 2*secondmax){
            return maxIndex;
        }
        return -1;
    }
}
```

### Python Solution

````python
class Solution:
    def dominantIndex(self, nums):
        max_num = -1
        second_max = -1
        max_index = -1

        for i in range(len(nums)):
            if nums[i] > max_num:
                second_max = max_num
                max_num = nums[i]
                max_index = i
            elif nums[i] > second_max:
                second_max = nums[i]

        if max_num >= 2 * second_max:
            return max_index

        return -1
````
