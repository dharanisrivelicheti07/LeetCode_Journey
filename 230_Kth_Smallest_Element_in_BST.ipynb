### 230. Kth Smallest Element in a BST
#### Given the root of a binary search tree, and an integer k, return the kth smallest value (1-indexed) of all the values of the nodes in the tree.
###### 

### Logic approach:
#### Set up: Create an empty stack to keep track of the nodes, and start a pointer at the root of the tree.Go as far left as possible: Loop through the tree, moving your pointer to the left child repeatedly. Push every node you pass onto the stack. (This ensures you reach the absolute smallest element in the current branch).Process the smallest available node: Once you can't go left anymore, pop the top node off the stack. This node is the next smallest element in the tree.Count it: Decrement your target number $k$ by 1 (or increment a separate counter).Check for success: If $k$ is now 0, congratulations! The value of the node you just popped is your answer.Move Right: If $k$ is not 0 yet, move your pointer to the right child of the node you just popped, and repeat the whole process from Step 2.

### Java Solution
```java
class Solution {
    public int kthSmallest(TreeNode root, int k) {
        Stack<TreeNode> stack = new Stack<>();
        TreeNode curr = root;
        while(curr!=null || !stack.isEmpty()){
            while(curr != null){
                stack.push(curr);
                curr = curr.left;
            }
            curr = stack.pop();
            k--;
            if(k==0){
                return curr.val;
            }
            curr = curr.right;
        }
        return -1;
    }
}
```

### Cpp Solution
```cpp
#include <stack>

class Solution {
public:
    int kthSmallest(TreeNode* root, int k) {
        std::stack<TreeNode*> stack;
        TreeNode* curr = root;
        
        while (curr != nullptr || !stack.empty()) {
           
            while (curr != nullptr) {
                stack.push(curr);
                curr = curr->left;
            }
            
            
            curr = stack.top();
            stack.pop();
            
           
            k--;
            if (k == 0) {
                return curr->val;
            }
            
            
            curr = curr->right;
        }
        
        return -1; // Fallback
    }
};
```

### Python solution

````python
class Solution:
    def kthSmallest(self, root: Optional[TreeNode], k: int) -> int:
        stack = []
        curr = root
        
        while curr or stack:
            
            while curr:
                stack.append(curr)
                curr = curr.left
                
            
            curr = stack.pop()
            
          
            k -= 1
            if k == 0:
                return curr.val
                
        
            curr = curr.right
            
        return -1
````
