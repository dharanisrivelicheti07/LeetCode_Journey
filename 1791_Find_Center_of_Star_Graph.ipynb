### 1791. Find Center of Star Graph
###### 

### Java Solution
```java
class Solution {
    public int findCenter(int[][] edges) {
        if(edges[0][0] == edges[1][0] || edges[0][0] == edges[1][1]){
            return edges[0][0];
        }
        return edges[0][1];
    }
}
```

### Cpp Solution
```cpp
class Solution {
public:
    int findCenter(vector<vector<int>>& edges) {
        
        // Check if the first node of the first edge
        // is present in the second edge
        if (edges[0][0] == edges[1][0] || 
            edges[0][0] == edges[1][1]) {
            
            return edges[0][0];
        }
        
        // Otherwise, the second node of the first edge
        // is the center
        return edges[0][1];
    }
};
```

### Python Solution

````python
class Solution:
    def findCenter(self, edges):
        
        
        if (edges[0][0] == edges[1][0] or 
            edges[0][0] == edges[1][1]):
            
            return edges[0][0]
        
        
        return edges[0][1]
````
