### 98. Validate Binary Search Tree
#### Given the root of a binary tree, determine if it is a valid binary search tree (BST).
#### A valid BST is defined as follows:
#### The left subtree of a node contains only nodes with keys strictly less than the node's key.
#### The right subtree of a node contains only nodes with keys strictly greater than the node's key.
#### Both the left and right subtrees must also be binary search trees.
###### 

### Java Solution
```java
class Solution {
    public boolean isValidBST(TreeNode root) {
        List<Integer> inOrderList = new LinkedList<>();
        helper(root,inOrderList);

        boolean isBST = true;
        int prev = inOrderList.get(0);
        for(int i=1;i<inOrderList.size();i++){
            if(inOrderList.get(i)<=prev)
            isBST = false;
            prev = inOrderList.get(i);
        }
        return isBST;
    }
    void helper(TreeNode treenode, List<Integer> inOrderList){
            if(treenode == null){
                return;
            }
            helper(treenode.left,inOrderList);
            inOrderList.add(treenode.val);
            helper(treenode.right,inOrderList);
        }
}
```

### Cpp Solution
```cpp
class Solution {
public:
    bool isValidBST(TreeNode* root) {
        vector<int> inOrderList;
        helper(root, inOrderList);

        if (inOrderList.empty()) return true;

        bool isBST = true;
        int prev = inOrderList[0];
        
        for (size_t i = 1; i < inOrderList.size(); i++) {
            if (inOrderList[i] <= prev) {
                isBST = false;
            }
            prev = inOrderList[i];
        }
        
        return isBST;
    }

private:
    void helper(TreeNode* treenode, vector<int>& inOrderList) {
        if (treenode == nullptr) {
            return;
        }
        helper(treenode->left, inOrderList);
        inOrderList.push_back(treenode->val);
        helper(treenode->right, inOrderList);
    }
};
```

### Python Solution

````python
class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        in_order_list = []
        self.helper(root, in_order_list)

        if not in_order_list:
            return True

        is_bst = True
        prev = in_order_list[0]

        for i in range(1, len(in_order_list)):
            if in_order_list[i] <= prev:
                is_bst = False
            prev = in_order_list[i]

        return is_bst

    def helper(self, treenode: Optional[TreeNode], in_order_list: list) -> None:
        if treenode is None:
            return
        
        self.helper(treenode.left, in_order_list)
        in_order_list.append(treenode.val)
        self.helper(treenode.right, in_order_list)
````
