### 111. Minimum Depth of Binary Tree
#### Given a binary tree, find its minimum depth.
#### The minimum depth is the number of nodes along the shortest path from the root node down to the nearest leaf node.
#### Note: A leaf is a node with no children
######

### Java Solution
```java
class Solution {
    public int minDepth(TreeNode root) {
        if(root == null){
            return 0;
        }
        Queue<TreeNode> queue =  new LinkedList<>();
        queue.offer(root);
        int depth = 1;
        while(!queue.isEmpty()){
            int levelSize = queue.size();
            for(int i=0;i<levelSize;i++){
                TreeNode currentNode = queue.poll();
                if(currentNode.left == null && currentNode.right == null){
                    return depth;
                }
                if(currentNode.left != null){
                    queue.offer(currentNode.left);
                }
                if(currentNode.right != null){
                    queue.offer(currentNode.right);
                }
            }
           depth++;
        }
        return depth;
    }
}
```

### Cpp Solution
```cpp
class Solution {
public:
    int minDepth(TreeNode* root) {
       
        if (!root) return 0;
        
        std::queue<TreeNode*> q;
        q.push(root);
        int depth = 1;
        
        while (!q.empty()) {
            int levelSize = q.size();
            
            
            for (int i = 0; i < levelSize; i++) {
                TreeNode* current = q.front();
                q.pop();
                
               
                if (!current->left && !current->right) {
                    return depth;
                }
                
                
                if (current->left) q.push(current->left);
                if (current->right) q.push(current->right);
            }
            
            depth++;
        }
        
        return depth;
    }
};
```

### Python Solution


````python
from collections import deque
from typing import Optional

class Solution:
    def minDepth(self, root: Optional[TreeNode]) -> int:
        # Handle edge case for an empty tree
        if not root:
            return 0
        
        queue = deque([root])
        depth = 1
        
        while queue:
            level_size = len(queue)
            
            # Process all nodes at the current depth level
            for _ in range(level_size):
                current = queue.popleft()
                
                # Early Exit: Check if it is a leaf node
                if not current.left and not current.right:
                    return depth
                    
                # Queue up existing children for the next level
                if current.left:
                    queue.append(current.left)
                if current.right:
                    queue.append(current.right)
                    
            # Increment depth after completely finishing the current level
            depth += 1
            
        return depth
````
